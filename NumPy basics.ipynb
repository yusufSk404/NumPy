***NumPy Data Types***

NumPy array

```python
#Method 1: From a Python List
arr1 = np.array([1, 2, 3, 4, 5])
print("From list:", arr1)

#Method 2: Using np.arange()
arr2 = np.arange(0, 10, 2)
print("Using arange:", arr2)

#Method 3: Using np.linspace()
arr3 = np.linspace(0, 1, 5)
print("Using linspace:", arr3)

#Method 4: Creating an Array of Zeros
zeros = np.zeros(5)
print("Zeros:", zeros)

#Method 5: Creating an Array of Ones
ones = np.ones(5)
print("Ones:", ones)

#Method 6: Creating an Array with a Specific Value Using np.full():
full = np.full((5, 7), 7)
print("Full (all 7s):", full)

#Method 7: Creating an Identity Matrix Using np.eye():
identity = np.eye(3)
print("Identity matrix:\n", identity)

#Method 8: Creating Random Arrays Using np.random.rand() and np.random.randint():
random_arr = np.random.rand(5)
print("Random (uniform):", random_arr)

random_int = np.random.randint(1, 10, size=5)
print("Random integers:", random_int)
```
